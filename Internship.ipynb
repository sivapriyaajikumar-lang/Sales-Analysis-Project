```python
plt.Figure(figsize=(5,5))
sns.countplot(data=df,x='Region',hue='Region',palette='viridis',legend=False)
plt.title('Sales count by Region ')
plt.xlabel('Region')
plt.ylabel('Number of Sales')
plt.show
```

This countplot shows the numbr of sales transactions for each regions. It helps to quickly identify which regions have lowest sales.
